```python
data = data[data['Monthly_Spend'] < 10000]
data
```

```python
data['Monthly_Spend_log'] = np.log1p(data['Monthly_Spend'])
data
```

```python
# Drop 'Engaged' and 'Customer_ID' from features
X = data.drop(columns=['Customer_ID', 'Engaged'])

# Target column
y = data['Engaged']
```

```python
X = sm.add_constant(X)
```

```python
model = sm.Logit(y, X).fit()


print(model.summary())
```
